# Teste 1 - Teste de conhecimento
Análise dos custos de transporte utilizando Python.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

dados = pd.read_csv("base_custos_transporte.csv")
dados
```

## Etapa 2 - Identificar as variáveis
**a)** Variável independente: mês.  
**b)** Variável dependente: custo de transporte.  
**c)** O mês representa o tempo e `custo_transporte` representa o valor gasto com transporte em cada mês.

```python
plt.scatter(dados['mes'], dados['custo_transporte'])
plt.title('Custos de transporte por mês')
plt.xlabel('Mês')
plt.ylabel('Custo de transporte (R$)')
plt.grid()
plt.show()
```

## Etapa 3 - Análise do gráfico
O gráfico mostra que os custos de transporte apresentam uma tendência de crescimento ao longo dos meses.

```python
# Etapa 4 - Encontrar a função f(x) = ax + b
a, b = np.polyfit(dados['mes'], dados['custo_transporte'], 1)

print('Valor de a:', round(a, 2))
print('Valor de b:', round(b, 2))
print(f'Função aproximada: f(x) = {a:.2f}x + {b:.2f}')
```

## Etapa 5 - Interpretar a função
O coeficiente **a** representa o aumento médio estimado do custo de transporte a cada mês.  
O coeficiente **b** representa o valor inicial estimado pela reta quando x = 0.  
Como **a** é positivo, os custos apresentam crescimento.

```python
# Etapa 6 - Dados reais e reta da função
reta = a * dados['mes'] + b

plt.scatter(dados['mes'], dados['custo_transporte'], label='Dados reais')
plt.plot(dados['mes'], reta, label='Função aproximada')
plt.title('Custos de transporte e tendência')
plt.xlabel('Mês')
plt.ylabel('Custo de transporte (R$)')
plt.legend()
plt.grid()
plt.show()
```

```python
# Etapa 7 - Previsões
custo_13 = a * 13 + b
custo_14 = a * 14 + b
custo_15 = a * 15 + b

print(f'Mês 13: f(13) = {a:.2f} * 13 + {b:.2f} = R$ {custo_13:.2f}')
print(f'Mês 14: f(14) = {a:.2f} * 14 + {b:.2f} = R$ {custo_14:.2f}')
print(f'Mês 15: f(15) = {a:.2f} * 15 + {b:.2f} = R$ {custo_15:.2f}')
```

## Etapa 8 - Análise dos resultados
**a)** Sim. A função representa bem a tendência geral de crescimento apresentada pelos dados.  
**b)** Os custos estão aumentando.  
**c)** Pela função encontrada, o custo aumenta aproximadamente **R$ 1.791,61 por mês**.

## Etapa 9 - Conclusão
Com base nos dados analisados, a empresa pode esperar que os custos de transporte continuem aumentando nos próximos meses. A função encontrada mostra um crescimento médio aproximado de R$ 1.791,61 por mês. Para os meses 13, 14 e 15, os valores estimados são aproximadamente R$ 65.495,45, R$ 67.287,06 e R$ 69.078,67.
